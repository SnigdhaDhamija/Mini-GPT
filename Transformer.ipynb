```python
import tensorflow as tf
import numpy as np
import re
from collections import Counter
```

```python
corpus = """
I love deep learning.
Deep learning is amazing.
I love transformers.
Transformers are powerful.
Deep learning powers GPT.
GPT generates text.
"""
```

```python
corpus = corpus.lower()
```

```python
corpus = re.sub(r"[^\w\s]","",corpus)
```

```python
tokens = corpus.split()
```

```python
word_counts = Counter(tokens)
```

```python
print(word_counts)
```

```python
vocab = {
    "<PAD>":0,
    "<UNK>":1
}
```

```python
for word in word_counts:
  vocab[word] = len(vocab)
```

```python
id_to_word = {id : word for word ,id in vocab.items()}
```

```python
encoded = [vocab.get(word,vocab["<UNK>"]) for word in tokens]
```

```python
print(encoded)
```

```python
inputs = []

targets = []
```

```python
for i in range(1,len(encoded)):
  input_sequence = encoded[:i]
  target = encoded[i]
  inputs.append(input_sequence)
  targets.append(target)
```

```python
print(inputs)
```

```python
print(targets)
```

```python
max_length = max(len(sequence) for sequence in inputs)
```

```python
print(max_length)
```

```python
pad_id = vocab["<PAD>"]
```

```python
padded_inputs = []

for sequence in inputs:

  if len(sequence) < max_length:

    sequence += [pad_id] * (max_length - len(sequence))

  padded_inputs.append(sequence)
```

```python
print(padded_inputs)
```

```python
input_tensor = tf.constant(
    padded_inputs, dtype = tf.int32
)
```

```python
target_tensor = tf.constant(
    targets, dtype = tf.int32
)
```

```python

print(input_tensor.shape)

print(target_tensor.shape)
```

```python
print(input_tensor.numpy())
```

```python
print(target_tensor.numpy())
```

```python
attention_mask = tf.cast(tf.not_equal(
    input_tensor,pad_id
)
,
tf.int32)
```

```python
dataset = tf.data.Dataset.from_tensor_slices(
    (
        {"input_ids":input_tensor,
         "attention_mask":attention_mask},
        target_tensor
    )
)
```

```python
dataset = dataset.shuffle(buffer_size = len(input_tensor))
```

```python
batch_size = 4

dataset = dataset.batch(batch_size)
```

```python
dataset = dataset.prefetch(
    tf.data.AUTOTUNE
)
```
